### 1️⃣ Mostrar los números pares del 1 al 20
**Problema:**
Usa un bucle `for` y `range()` para imprimir todos los números **pares** entre 1 y 20 (ambos incluidos).

Pista: utiliza la condición `if` dentro del bucle.

````python
# 💻 Intenta resolverlo aquí
def pares_hasta_20():
    pass
````

````python
# ✅ Solución sugerida
def pares_hasta_20():
    return [n for n in range(1, 21) if n % 2 == 0]
````

### 2️⃣ Contar vocales y consonantes
**Problema:**
Escribe una función `contar_letras(cadena)` que devuelva el número de vocales y consonantes de una cadena.

Ejemplo:
```python
contar_letras('hola')  # (2, 2)
```

````python
def contar_letras(cadena):
    pass
````

````python
def contar_letras(cadena):
    vocales = 'aeiouAEIOU'
    num_vocales, num_consonantes = 0, 0
    for c in cadena:
        if c in vocales:
            num_vocales += 1
        elif c.isalpha():
            num_consonantes += 1
    return num_vocales, num_consonantes
````

````python
assert contar_letras('hola') == (2, 2)
assert contar_letras('adios1') == (3, 2)
````

## 3️⃣ Calcular la suma de los múltiplos de 3 entre 1 y 30
**Problema:**
Usa `for` y `range()` para sumar todos los múltiplos de 3 entre 1 y 30.

Ejemplo:
```python
suma_multiplos_3()  # 165
```

````python
def suma_multiplos_3():
    pass
````

````python
def suma_multiplos_3():
    total = 0
    for n in range(1, 31):
        if n % 3 == 0:
            total += n
    return total
````

````python
assert suma_multiplos_3() == 165
````

## 4️⃣ Detectar número mayor en una lista sin usar `max()`
**Problema:** Implementa `mayor(lista)` que devuelva el número más grande de una lista.

````python
def mayor(lista):
    pass
````

````python
def mayor(lista):
    maximo = lista[0]
    for n in lista:
        if n > maximo:
            maximo = n
    return maximo
````

````python
assert mayor([1,5,3,9,2]) == 9
````

## 5️⃣ Uso de `break` para detener un bucle
**Problema:** Crea una función `buscar_numero(lista, valor)` que recorra una lista y se detenga cuando encuentre el número buscado, devolviendo `True` si lo encuentra.

Utiliza `break`.

````python
def buscar_numero(lista, valor):
    pass
````

````python
def buscar_numero(lista, valor):
    encontrado = False
    for n in lista:
        if n == valor:
            encontrado = True
            break
    return encontrado
````

````python
assert buscar_numero([2,4,6,8],6)
assert not buscar_numero([1,3,5],2)
````

## 6️⃣ Uso de `continue` para saltar números impares
**Problema:** Crea una función `pares(lista)` que devuelva una lista con solo los números pares, usando `continue` para saltar los impares.

````python
def pares(lista):
    pass
````

````python
def pares(lista):
    resultado = []
    for n in lista:
        if n % 2 != 0:
            continue
        resultado.append(n)
    return resultado
````

````python
assert pares([1,2,3,4,5,6]) == [2,4,6]
````

## 7️⃣ Clasificar edad usando `match`
**Problema:** Usa `match` para devolver un mensaje según el rango de edad.

Ejemplo:
```python
clasificar_edad(10) → 'Niño'
```

````python
def clasificar_edad(edad):
    pass
````

````python
def clasificar_edad(edad):
    match edad:
        case n if n < 12:
            return 'Niño'
        case n if n < 18:
            return 'Adolescente'
        case _:
            return 'Adulto'
````

````python
assert clasificar_edad(10) == 'Niño'
assert clasificar_edad(16) == 'Adolescente'
assert clasificar_edad(30) == 'Adulto'
````

## 8️⃣ Clasificar día de la semana con `match`
**Problema:** Usa `match` para devolver el tipo de día según su nombre:

- Lunes a Viernes → 'Laboral'
- Sábado y Domingo → 'Fin de semana'

````python
def tipo_dia(dia):
    pass
````

````python
def tipo_dia(dia):
    match dia.lower():
        case 'lunes' | 'martes' | 'miércoles' | 'jueves' | 'viernes':
            return 'Laboral'
        case 'sábado' | 'domingo':
            return 'Fin de semana'
        case _:
            return 'Día no válido'
````

````python
assert tipo_dia('martes') == 'Laboral'
assert tipo_dia('domingo') == 'Fin de semana'
assert tipo_dia('fiesta') == 'Día no válido'
````
